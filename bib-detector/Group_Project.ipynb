```python
import cv2

model = ultralytics.YOLO("/content/drive/MyDrive/University/Group Project/Running/best.pt")

input_video = "/content/drive/MyDrive/University/Group Project/Running/drogo.mp4"
output_video = "/content/drive/MyDrive/University/Group Project/Running/output.mp4"

cap = cv2.VideoCapture(input_video)

frame_width = int(cap.get(3))
frame_height = int(cap.get(4))
fps = int(cap.get(cv2.CAP_PROP_FPS))

out = cv2.VideoWriter(output_video, cv2.VideoWriter_fourcc(*'mp4v'), fps, (frame_width, frame_height))

if not cap.isOpened():
  print("NO FILE")
  exit()

while cap.isOpened():
  ret, frame = cap.read()
  if not ret:
    break
  results = model(frame, device=0)
  for result in results:
    for box in result.boxes:
      x1, y1, x2, y2 = map(int, box.xyxy[0])
      cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

      conf = float(box.conf[0])
      label = result.names[int(box.cls[0])]

      cv2.putText(frame, f"{label}: {conf:.2f}", (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
  out.write(frame)

cap.release()
out.release()
cv2.destroyAllWindows()
```
